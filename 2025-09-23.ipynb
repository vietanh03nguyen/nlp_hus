```scala
val path = "../data/c4_30k/c4-train.00000-of-01024-30K.json"



val df = spark.read.json(path)

df.show(5)

df.count

val ef = df.select("text")

ef.show(5)

ef.row

```

```scala
import org.apache.spark.ml.feature._
val tokenizer = new Tokenizer().setInputCol("text").setOutputCol("tokens")
val ff = tokenizer.transform(ef)
ff.show(5)
ff.printSchema
ff.select("tokens").head
```

```scala
val vectorizer = new CountVectorizer().setInputCol("tokens").setOutputCol("vector").setMinDF(3) #Min Document Frequency: Xuất hiện ở 3 tài liệu khác nhau != Xuất hiện 3 lần
val model = vectorizer.fit(ff)
# val vectorizer = new CountVectorizer().setInputCol("tokens").setOutputCol("vector")
model.vocabulary
model.vocabulary.size
model.vocabulary.takeRight(20)
val gf = model.transform(ff)

gf.show(5)
gf.printSchema
gf.select("vector").head # Chứa 2 mảng: mảng index và mảng frequency của từ tại index tương ứng

```
